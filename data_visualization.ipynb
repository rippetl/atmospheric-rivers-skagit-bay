```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#reads in all casts and puts them in a dataframe
files = ['cast1.csv', 'cast2.csv', 'cast3.csv', 'cast4.csv', 'cast5.csv', 'cast6.csv', 'cast7.csv', 'cast8.csv', 'cast9.csv', 'cast10.csv', 'cast11.csv', 'cast12.csv']
cast_list = [pd.read_csv('/Users/lilyrippeteau/Library/CloudStorage/OneDrive-WesternWashingtonUniversity/Shortcuts/WWU Environmental Hydrodynamics Lab - General/Atmospheric Rivers/' +f) for f in files]
```

```python
#loops through each cast and plots salinity vs pressure and depth
for i, cast1 in enumerate(cast_list):
    cast1_pressure = cast1[['Pressure']]
    cast1_depth = cast1[['Depth']]
    cast1_salinity = cast1[['Salinity']]

    fig, axes = plt.subplots(1, 2)

    axes[0].plot(cast1_salinity, cast1_pressure, color="red")
    axes[0].set_xlabel('Salinity (psu)')
    axes[0].set_ylabel('Pressure (dbar)')
    axes[0].invert_yaxis()
    axes[0].set_title("Salinity vs Pressure (dbar) at Cast " + str(i+1))

    axes[1].plot(cast1_salinity, cast1_depth, color="red")
    axes[1].set_xlabel('Salinity (psu)')
    axes[1].set_ylabel('Depth (m)')
    axes[1].invert_yaxis()
    axes[1].set_title("Salinity vs Depth (m) at Cast " + str(i+1))

    plt.tight_layout()
    plt.show()
```
